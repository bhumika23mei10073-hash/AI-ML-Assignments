```python
# Movie Recommendation System
# Simple Content-Based Filtering

import pandas as pd

# Small dataset
movies = {
    "title": [
        "Avengers",
        "Iron Man",
        "Spider-Man",
        "Titanic",
        "The Notebook",
        "Interstellar",
        "Inception",
        "The Dark Knight",
        "Toy Story",
        "Finding Nemo"
    ],

    "genre": [
        "Action Adventure",
        "Action Adventure",
        "Action Adventure",
        "Romance Drama",
        "Romance Drama",
        "Sci-Fi Adventure",
        "Sci-Fi Action",
        "Action Crime",
        "Animation Comedy",
        "Animation Adventure"
    ]
}

df = pd.DataFrame(movies)

print("Movie Dataset:")
display(df)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Convert genres into numbers
vectorizer = CountVectorizer()

genre_matrix = vectorizer.fit_transform(df["genre"])

# Calculate similarity between movies
similarity = cosine_similarity(genre_matrix)

print("Recommendation model created successfully!")
```

```python
def recommend_movie(movie_name, number_of_movies=5):

    if movie_name not in df["title"].values:
        print("Movie not found!")
        return

    movie_index = df[df["title"] == movie_name].index[0]

    # Get similarity scores
    scores = list(enumerate(similarity[movie_index]))

    # Sort from highest to lowest
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    print("Movies similar to:", movie_name)
    print()

    count = 0

    for index, score in scores[1:]:
        print(df.iloc[index]["title"], "-> Similarity:", round(score, 2))

        count += 1

        if count == number_of_movies:
            break
```

```python
recommend_movie("Avengers")
```

```python
recommend_movie("Titanic")
```
