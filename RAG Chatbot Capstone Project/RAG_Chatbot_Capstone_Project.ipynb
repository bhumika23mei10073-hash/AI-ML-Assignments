```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("AI model loaded successfully!")
```

```python
prompt = "Explain artificial intelligence in simple words."

inputs = tokenizer(
    prompt,
    return_tensors="pt",
    truncation=True
)

outputs = model.generate(
    **inputs,
    max_new_tokens=100
)

answer = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Answer:", answer)
```

```python
def generate_answer(question, context):
    prompt = f"""
Answer the question using the context below.

Context:
{context}

Question:
{question}

Answer:
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=100
    )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return answer
```

```python
context = """
Artificial Intelligence is a branch of computer science.
It allows computers to perform tasks that normally require human intelligence.
"""

question = "What is Artificial Intelligence?"

answer = generate_answer(question, context)

print("Question:", question)
print("Answer:", answer)
```

```python
document = """
Artificial Intelligence (AI) is a branch of computer science that focuses on
creating machines that can perform tasks requiring human intelligence.

Machine Learning (ML) is a subset of AI that allows computers to learn from
data without being explicitly programmed.

Deep Learning is a subset of Machine Learning that uses neural networks with
multiple layers to learn complex patterns.

Natural Language Processing (NLP) is a field of AI that helps computers
understand and generate human language.

A chatbot is a software application that communicates with users using
natural language.
"""

with open("knowledge.txt", "w") as file:
    file.write(document)

print("Knowledge document created successfully!")
```

```python
with open("knowledge.txt", "r") as file:
    text = file.read()

print(text)
```

```python
def split_text(text, chunk_size=300):
    chunks = []

    for i in range(0, len(text), chunk_size):
        chunks.append(text[i:i + chunk_size])

    return chunks

chunks = split_text(text)

print("Number of chunks:", len(chunks))

for i, chunk in enumerate(chunks):
    print("\nChunk", i + 1)
    print(chunk)
```

```python
def retrieve_context(question, chunks):
    question_words = question.lower().split()

    best_chunk = ""
    best_score = 0

    for chunk in chunks:
        score = sum(
            1 for word in question_words
            if word in chunk.lower()
        )

        if score > best_score:
            best_score = score
            best_chunk = chunk

    return best_chunk
```

```python
def rag_chatbot(question):
    context = retrieve_context(question, chunks)

    prompt = f"""
Use the following context to answer the question.

Context:
{context}

Question:
{question}

Answer:
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=100
    )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return answer
```

```python
question = input("Ask your question: ")

answer = rag_chatbot(question)

print("\nChatbot Answer:")
print(answer)
```

```python
while True:
    question = input("\nYou: ")

    if question.lower() in ["exit", "quit", "bye"]:
        print("Chatbot: Goodbye!")
        break

    answer = rag_chatbot(question)

    print("Chatbot:", answer)
```

```python
from pypdf import PdfReader

reader = PdfReader(pdf_name)

pdf_text = ""

for page in reader.pages:
    text = page.extract_text()

    if text:
        pdf_text += text + "\n"

print("PDF text extracted successfully!")
print("Characters:", len(pdf_text))
```

```python
def rag_chatbot(question):

    context = retrieve_context(question, chunks)

    prompt = f"""
Answer the question using only the information provided in the context.

Context:
{context}

Question:
{question}

Answer:
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=100
    )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return answer
```
